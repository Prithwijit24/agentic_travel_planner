## 5. Retrieval System

### 5a. Chunker (`retrieval/chunker.py`)
Splits text into fixed-size chunks with configurable overlap. Pure function, no dependencies.

````python
from agentic_tour_planner.retrieval.chunker import chunk_text

sample = "Paris is the capital of France. " * 50
chunks = list(chunk_text(sample, chunk_size=200, overlap=30))
print(f"{len(sample)} chars → {len(chunks)} chunks")
for i, c in enumerate(chunks[:3]):
    print(f"  Chunk {i}: {len(c)} chars — {c[:50]}...")
````

### 5b. VectorStore (`retrieval/vector_store.py`)
Two modes:
- **ChromaDB + FastEmbed** — persistent vector DB with local embeddings (BGE models)
- **Fallback** — in-memory dict with bag-of-words scoring (no external deps)

`upsert_documents(docs)` → chunks + embeds + stores.
`retrieve(query, top_k)` → semantic + fallback search.

````python
from agentic_tour_planner.retrieval.vector_store import VectorStore
from agentic_tour_planner.domain.models import SourceDocument

vs = VectorStore()

# Ingest a sample document
doc = SourceDocument(
    source_id="test:paris",
    source_type="web",
    title="Paris Travel Guide",
    content="Paris has many museums including the Louvre and Musée d'Orsay. "
            "The Eiffel Tower is a famous landmark. French cuisine is renowned worldwide.",
)
n = vs.upsert_documents([doc])
print(f"Indexed {n} chunks")

# Retrieve
results = vs.retrieve("museums in Paris", top_k=3)
for r in results:
    print(f"  [{r.source_type}] {r.title}: {r.content[:80]}...")
````

### 5c. Reranker (`retrieval/reranker.py`)
`HybridReranker` — combines vector score + lexical overlap (token overlap with query).
Acts as a second-pass filter over initial retrieval results.

````python
from agentic_tour_planner.retrieval.reranker import HybridReranker

reranker = HybridReranker()
reranked = reranker.rerank("museums in Paris", results, top_k=2)
for r in reranked:
    print(f"  Score: {r.metadata.get('score', 0):.3f} — {r.content[:60]}...")
````

### 8b. Graph (`pipeline/graph.py`)
LangGraph state machine with 4 sequential nodes:
1. `gather_context` — async retrieval + tool calls
2. `build_insights` — route/budget/timing analysis
3. `build_prompt` — assemble the prompt
4. `generate_plan` — LLM JSON generation

### 8c. AgenticTourPlannerPipeline (`pipeline/agentic_pipeline.py`)
Wraps the graph with convenience methods:
- `gather_context()` — single-step retrieval + tool orchestration
- `run()` — full end-to-end: graph → parse → build `PlanningResponse`

````python
from agentic_tour_planner.pipeline.agentic_pipeline import AgenticTourPlannerPipeline
from agentic_tour_planner.pipeline.graph import build_planner_graph

# Build the graph standalone
graph = build_planner_graph(
    gather_context=lambda req: asyncio.run(AgenticTourPlannerPipeline().gather_context(req)),
    llm_provider=provider,
    insights_builder=builder,
)
print(f"Graph built: {type(graph).__name__}")

# Pipeline wrapper
pipeline = AgenticTourPlannerPipeline(provider=provider)
print(f"Pipeline ready — gather_context, run available")
````

## 9. End-to-End Pipeline Run

Runs the full pipeline: gather context → insights → prompt → LLM → response.
Uses the **fallback JSON generator** when no API key is available.

````python
request = PlanningRequest(
    destination="Tokyo",
    origin="San Francisco",
    trip_length_days=2,
    interests=["technology", "food", "temples"],
    budget_level="midrange",
    travel_month="October",
    notes="Interested in robot cafes",
    include_live_data=False,
)

response = await pipeline.run(request)

print(f"Plan ID: {response.plan_id}")
print(f"Provider: {response.provider_used}/{response.model_used}")
print(f"Generated: {response.generated_at}")
print(f"\nOverview: {response.overview}")
print(f"\nItinerary ({len(response.itinerary)} days):")
for day in response.itinerary:
    print(f"\n  Day {day.day}: {day.theme}")
    print(f"    Morning: {', '.join(day.morning)[:80]}...")
    print(f"    Afternoon: {', '.join(day.afternoon)[:80]}...")
    print(f"    Evening: {', '.join(day.evening)[:80]}...")
print(f"\nTips ({len(response.practical_tips)}):")
for t in response.practical_tips:
    print(f"  • {t}")
print(f"\nCitations ({len(response.citations)}): " + ', '.join(c.title for c in response.citations[:3]))
print(f"\nInsights:")
pprint(response.insights.model_dump())
````

## 12. Evaluation (RAGAS)

`evaluation/ragas_pipeline.py` — `RagasEvaluationPipeline`:
- `load_cases(path)` — loads evaluation cases from JSON (supports legacy formats)
- `build_dataset_rows(cases_path)` — runs pipeline for each case
- `export_dataset()` — saves dataset as JSON
- `run_ragas()` — evaluates using RAGAS metrics (answer_relevancy, faithfulness, context_recall)

Graceful degradation: if `ragas`/`datasets` packages are missing, computes heuristic scores.

````python
from agentic_tour_planner.evaluation.ragas_pipeline import RagasEvaluationPipeline

eval_pipeline = RagasEvaluationPipeline()

# Check for eval dataset
eval_dir = settings.evaluation_dir
eval_dir.mkdir(parents=True, exist_ok=True)

# Create a sample eval case inline
cases = [
    {
        "question": "What are the best museums to visit in Paris for art lovers?",
        "ground_truth": "The Louvre and Musée d'Orsay are the top art museums in Paris.",
        "reference_contexts": ["Louvre Museum", "Musée d'Orsay"],
        "request": {"destination": "Paris", "trip_length_days": 3, "interests": ["art"]},
    }
]
cases_file = eval_dir / "test_cases.json"
cases_file.write_text(json.dumps(cases, indent=2))

loaded = eval_pipeline.load_cases(cases_file)
print(f"Loaded {len(loaded)} evaluation case(s):")
for c in loaded:
    print(f"  Q: {c.question[:60]}...")
    print(f"  GT: {c.ground_truth[:60]}...")
````

````python
# ── Export dataset ──
try:
    report = await eval_pipeline.export_dataset(cases_file)
    print(f"Dataset exported: {report.output_path}")
    print(f"Metrics: {json.dumps(report.metrics, indent=2)}")
except Exception as e:
    print(f"Export error (expected if LLM unavailable): {e}")
````

````python
# ── Run RAGAS evaluation ──
try:
    report = await eval_pipeline.run_ragas(cases_file)
    print(f"RAGAS report: {report.output_path}")
    print(f"Metrics:")
    for k, v in report.metrics.items():
        print(f"  {k}: {v}")
except Exception as e:
    print(f"RAGAS evaluation error: {e}")
````
